# Kruskal's algo 

Kruskal’s algorithm is another greedy algorithm, but instead of growing a single tree from a vertex, it builds the MST by considering edges globally.

How it works:
1. Sort all the edges in the graph in non-decreasing order of their weight.

2. Iterate through the sorted edges one by one. 

3. For each edge, check if adding it creates a cycle using a Disjoint Set Union (DSU) / Union-Find data structure.

4. If it does not form a cycle, include it in the MST and merge the sets. Stop when the MST contains $V - 1$ edges.

```python
class DisjointSet:

  def __init__(self, n):
    self.parent = list(range(n))
    self.rank = [0] * n

  def find(self, i):
    if self.parent[i] == i:
      return i
    self.parent[i] = self.find(self.parent[i])  # Path compression
    return self.parent[i]

  def union(self, i, j):
    root_i = self.find(i)
    root_j = self.find(j)

    if root_i != root_j:
      # Union by rank
      if self.rank[root_i] < self.rank[root_j]:
        self.parent[root_i] = root_j
      elif self.rank[root_i] > self.rank[root_j]:
        self.parent[root_j] = root_i
      else:
        self.parent[root_j] = root_i
        self.rank[root_i] += 1
      return True
    return False


def kruskals_algorithm(n, edges):
  """n: number of vertices

  edges: list of tuples in the format (weight, u, v)
  """
  edges.sort()  # Sort edges by weight
  ds = DisjointSet(n)
  mst_weight = 0
  mst_edges = []

  for weight, u, v in edges:
    if ds.union(u, v):
      mst_weight += weight
      mst_edges.append((u, v, weight))

  return mst_weight, mst_edges


# Example Usage:
if __name__ == "__main__":
  # Graph with 4 vertices (0, 1, 2, 3)
  # Edges list: (weight, u, v)
  n = 4
  edges = [(1, 0, 1), (3, 0, 2), (1, 1, 2), (4, 1, 3), (2, 2, 3)]

  weight, mst_edges = kruskals_algorithm(n, edges)
  print("\n--- Kruskal's Algorithm ---")
  print(f"Total MST Weight: {weight}")
  print(f"Edges in MST (u, v, weight): {mst_edges}")
```
