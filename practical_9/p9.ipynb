# Prim's Algo
Explanation
Prim’s algorithm is a greedy algorithm that builds the MST starting from an arbitrary single vertex and expanding outward, one vertex at a time.

How it works:

Maintain two sets of vertices: those included in the MST and those not yet included.

Start with an arbitrary vertex (added to the MST).

At each step, find all edges connecting a vertex inside the MST to a vertex outside the MST.

Pick the edge with the minimum weight, add the new vertex to the MST, and repeat until all vertices are included.

Data Structure: Typically implemented using a Min-Priority Queue (Heap) to efficiently find the minimum weight edge.

```python
import heapq


def prims_algorithm(n, adj):
  """n: number of vertices (0 to n-1)

  adj: adjacency list where adj[u] contains tuples of (weight, v)
  """
  visited = [False] * n
  min_heap = [(0, 0)]  # (weight, current_vertex)
  mst_weight = 0
  mst_edges = []

  while min_heap:
    weight, u = heapq.heappop(min_heap)

    if visited[u]:
      continue

    visited[u] = True
    mst_weight += weight

    # Skip recording the initial dummy edge (0, 0)
    if weight > 0:
      mst_edges.append((weight, u))

    for v, w in adj[u]:
      if not visited[v]:
        heapq.heappush(min_heap, (w, v))

  return mst_weight, mst_edges


# Example Usage:
if __name__ == "__main__":
  # Graph with 4 vertices (0, 1, 2, 3)
  # Adjacency list: vertex -> list of (neighbor, weight)
  n = 4
  adj = {
      0: [(1, 1), (2, 3)],
      1: [(0, 1), (2, 1), (3, 4)],
      2: [(0, 3), (1, 1), (3, 2)],
      3: [(1, 4), (2, 2)],
  }

  weight, edges = prims_algorithm(n, adj)
  print("--- Prim's Algorithm ---")
  print(f"Total MST Weight: {weight}")
  print(f"Edges in MST (Weight, Vertex): {edges}")
```
